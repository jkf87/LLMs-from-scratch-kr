## 연습 7.1: 프롬프트 스타일 변경

다음과 같은 데이터 항목이 있다고 가정해봅시다:

```json
{
  "instruction": "Identify the correct spelling of the following word.",
  "input": "Ocassion",
  "output": "The correct spelling is 'Occasion.'"
}
```

메인 장에서는 이것을 Alpaca 스타일 프롬프트 템플릿에 따라 다음과 같이 형식화했습니다:

```
Below is an instruction that describes a task. Write a response that appropriately completes the request.

### Instruction:
Identify the correct spelling of the following word.

### Input:
Occassion

### Response:
The correct spelling is 'Occasion.'
```

이번 연습에서는 대신 Phi-3 프롬프트 템플릿을 사용하여 데이터 항목을 다음과 같이 형식화합니다:

```
<user>
Identify the correct spelling of the following word: 'Occasion'

<assistant>
The correct spelling is 'Occasion'.
```

이 프롬프트 템플릿은 상당히 짧아서 입력 프롬프트가 짧아지므로 LLM 미세조정과 텍스트 생성의 실행 시간과 하드웨어 요구 사항을 줄입니다.
이 변경을 위해 `format_input` 함수를 다음과 같이 업데이트합니다:

````python
def format_input(entry):
    instruction_text = (
        f"<|user|>\n{entry['instruction']}"
    )

    input_text = f"\n{entry['input']}" if entry["input"] else ""

    return instruction_text + input_text
````

`'input'` 필드에 내용이 있는 것과 없는 것, 두 개의 입력 샘플에 적용하여 의도한 대로 작동하는지 확인해봅시다:

````python
sample_data = [
    {'instruction': 'Identify the correct spelling of the following word.', 'input': 'Ocassion', 'output': "The correct spelling is 'Occasion.'"}, 
    {'instruction': "What is an antonym of 'complicated'?", 'input': '', 'output': "An antonym of 'complicated' is 'simple'."}
]

print(format_input(sample_data[0]))
print()
print(format_input(sample_data[1]))
````

#### 팁: 특수 토큰 고려사항

- Phi-3 프롬프트 템플릿에는 `<|user|>`와 `<|assistant|>`와 같은 특수 토큰이 포함되어 있으며, 이는 GPT-2 토크나이저에게는 최적이 아닐 수 있습니다
- GPT-2 토크나이저는 `<|endoftext|>`를 특수 토큰(토큰 ID 50256으로 인코딩)으로 인식하지만, 앞서 언급한 다른 특수 토큰을 처리하는 데는 비효율적입니다
- 예를 들어, `<|user|>`는 5개의 개별 토큰 ID(27, 91, 7220, 91, 29)로 인코딩되어 매우 비효율적입니다
- `allowed_special` 인수를 통해 `tiktoken`에서 `<|user|>`를 새로운 특수 토큰으로 추가할 수 있지만, GPT-2 어휘는 추가 수정 없이는 이를 처리할 수 없다는 점을 기억하세요
- 토크나이저와 LLM을 확장하여 특수 토큰을 처리하는 방법에 대해 궁금하다면, [extend-tiktoken.ipynb](../../ch05/09_extending-tokenizers/extend-tiktoken.ipynb) 보너스 자료를 참조하세요(여기서는 필수가 아니지만 호기심 많은 독자들을 위한 흥미로운/보너스 고려사항입니다)
- 또한, 어휘를 통해 프롬프트 템플릿의 이러한 특수 토큰을 지원하는 모델이 더 효율적이고 전반적으로 더 나은 성능을 보일 수 있다고 가정할 수 있습니다

## 연습 7.2: 지시사항 및 입력 마스킹

다음 그림에 표시된 대로 지시사항을 마스킹하려면 `InstructionDataset` 클래스와 `custom_collate_fn`을 약간 수정해야 합니다.

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/ch07_compressed/mask-instructions.webp" width=600px>

````python
# 이 `format_input` 함수는 원본 7장 코드에서 복사했습니다

def format_input(entry):
    instruction_text = (
        f"Below is an instruction that describes a task. "
        f"Write a response that appropriately completes the request."
        f"\n\n### Instruction:\n{entry['instruction']}"
    )

    input_text = f"\n\n### Input:\n{entry['input']}" if entry["input"] else ""

    return instruction_text + input_text
````

콜레이트 함수를 코딩할 때 대상에서 지시사항 내용 위치를 찾기 위해 사용할 지시사항의 길이를 수집하도록 `InstructionDataset` 클래스를 다음과 같이 수정할 수 있습니다:

````python
import torch
from torch.utils.data import Dataset


class InstructionDataset(Dataset):
    def __init__(self, data, tokenizer):
        self.data = data

        ##########################################################################################
        # 새로운 내용: 지시사항 길이를 위한 별도 리스트
        self.instruction_lengths = []
        ##########################################################################################
        
        self.encoded_texts = []
        
        for entry in data:
            instruction_plus_input = format_input(entry)
            response_text = f"\n\n### Response:\n{entry['output']}"
            full_text = instruction_plus_input + response_text
            
            self.encoded_texts.append(
                tokenizer.encode(full_text)
            )

            ##########################################################################################
            # 새로운 내용: 지시사항 길이 수집
            instruction_length = len(tokenizer.encode(instruction_plus_input))
            self.instruction_lengths.append(instruction_length)
            ##########################################################################################
            
    def __getitem__(self, index):
        # 새로운 내용: 지시사항 길이와 텍스트를 별도로 반환
        return self.instruction_lengths[index], self.encoded_texts[index]

    def __len__(self):
        return len(self.data)
````

````python
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")
````

다음으로, `InstructionDataset` 데이터셋의 변경으로 인해 각 `batch`가 이제 단순한 `item` 대신 `(instruction_length, item)`을 포함하는 튜플이 된 `custom_collate_fn`을 업데이트합니다. 또한 이제 대상 ID 리스트에서 해당 지시사항 토큰을 마스킹합니다.

````python
def custom_collate_fn(
    batch,
    pad_token_id=50256,
    ignore_index=-100,
    allowed_max_length=None,
    device="cpu"
):
    # 배치에서 가장 긴 시퀀스 찾기
    batch_max_length = max(len(item)+1 for instruction_length, item in batch)   # 새로운 내용: batch가 이제 튜플

    # 입력과 대상 패딩 및 준비
    inputs_lst, targets_lst = [], []

    for instruction_length, item in batch:  # 새로운 내용: batch가 이제 튜플
        new_item = item.copy()
        # <|endoftext|> 토큰 추가
        new_item += [pad_token_id]
        # 시퀀스를 max_length로 패딩
        padded = new_item + [pad_token_id] * (batch_max_length - len(new_item))
        inputs = torch.tensor(padded[:-1])  # 입력용으로 마지막 토큰 자르기
        targets = torch.tensor(padded[1:])  # 대상용으로 +1 오른쪽으로 시프트

        # 대상에서 첫 번째를 제외한 모든 패딩 토큰을 ignore_index로 교체
        mask = targets == pad_token_id
        indices = torch.nonzero(mask).squeeze()
        if indices.numel() > 1:
            targets[indices[1:]] = ignore_index

        ##########################################################################################
        # 새로운 내용: 대상에서 모든 입력 및 지시사항 토큰 마스킹
        targets[:instruction_length-1] = -100
        ##########################################################################################
        
        # 선택적으로 최대 시퀀스 길이로 자르기
        if allowed_max_length is not None:
            inputs = inputs[:allowed_max_length]
            targets = targets[:allowed_max_length]
        
        inputs_lst.append(inputs)
        targets_lst.append(targets)

    # 입력 및 대상 리스트를 텐서로 변환하고 대상 장치로 전송
    inputs_tensor = torch.stack(inputs_lst).to(device)
    targets_tensor = torch.stack(targets_lst).to(device)

    return inputs_tensor, targets_tensor
````

아래의 샘플 데이터에서 테스트해봅시다:

````python
sample_data = [
    {'instruction': "What is an antonym of 'complicated'?", 'input': '', 'output': "An antonym of 'complicated' is 'simple'."},
    {'instruction': 'Sort the following list in alphabetical order.', 'input': 'Zebra, Elephant, Crocodile', 'output': 'Crocodile, Elephant, Zebra'},
    {'instruction': 'Arrange the given numbers in descending order.', 'input': '5, 12, 8, 3, 15', 'output': '15, 12, 8, 5, 3.'}
]
````

````python
from torch.utils.data import DataLoader

train_dataset = InstructionDataset(sample_data, tokenizer)
train_loader = DataLoader(
    train_dataset,
    batch_size=len(sample_data),
    collate_fn=custom_collate_fn,
    num_workers=0
)
````

````python
print("Train loader:")
for inputs, targets in train_loader:
    print(inputs.shape, targets.shape)
````
